```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error, r2_score
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/mpg.csv"

df=pd.read_csv(url)

df.head()
```

```python
X = df[['displacement']]
Y = df['mpg']
print("Missing values in displacement:", X.isnull().sum())
print("Missing values in mpg:", Y.isnull().sum())
```

```python
data = df[['displacement','mpg']].dropna()

X = data[['displacement']]
Y = data['mpg']

print("Feature:")
print(X.head())

print("Target:")
print(Y.head())
```

```python
plt.figure(figsize=(8,5))

plt.scatter(X,Y)
plt.xlabel("Engine Displacement")
plt.ylabel("Miles Per Gallon(mpg)")
plt.title("Engine Displacement vs MPG")

plt.grid(True)
plt.show()
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42
)

print("Training samples:",len(X_train))
print("Testing samples:",len(X_test))

    
```

```python
linear_model = LinearRegression()

linear_model.fit(X_train ,Y_train)

Y_pred_linear = linear_model.predict(X_test)
```

```python
mse_linear = mean_squared_error(Y_test,Y_pred_linear)
r2_linear = r2_score(Y_test,Y_pred_linear)

print("Linera Regression Results")
print("--------------------------")
print("MSE: ",mse_linear)
print("R2 Score: ",r2_linear)
```

```python
results = []
polynomial_models = {}

results.append({
   'Model': 'Linear Regression',
    'Degree': 1,
    'MSE': mse_linear,
    'R-squared': r2_linear
})
for degree in range(2,6):
    poly = PolynomialFeatures(degree=degree)

    X_train_poly=poly.fit_transform(X_train)
    X_test_poly=poly.transform(X_test)

    poly_model=LinearRegression()
    poly_model.fit(X_train_poly,Y_train)
    Y_pred_poly = poly_model.predict(X_test_poly)

    mse = mean_squared_error(Y_test,Y_pred_poly)
    r2 = r2_score(Y_test,Y_pred_poly)

    results.append({
       'Model': 'Polynomial Regression',
       'Degree': degree,
       'MSE': mse,
       'R-squared': r2
    })

    polynomial_models[degree] = {
       'poly':poly,
       'model':poly_model
    }
```

```python
results_df = pd.DataFrame(results)
results_df
```

```python
X_plot = pd.DataFrame({
    'displacement':np.linspace(
        X['displacement'].min(),
        X['displacement'].max(),
        30)
    
})

plt.figure(figsize=(10,6))

plt.scatter(
   X['displacement'],
   Y,
    label='Actual Data'
)

Y_plot_linear = linear_model.predict(X_plot)

plt.plot(
    X_plot['displacement'],
    Y_plot_linear,
    label='Linear Regression'
)

for degree in range(2,6):
    poly=polynomial_models[degree]['poly']
    poly_model = polynomial_models[degree]['model']

    X_plot_poly = poly.transform(X_plot)

    Y_plot_poly = poly_model.predict(X_plot_poly)

    plt.plot(
       X_plot['displacement'],
       Y_plot_poly,
       label=f'Polynomial Degree {degree}'
    )

plt.xlabel('Engine Displacement')
plt.ylabel('Miles Per Gallon')

plt.title('Linear and Polynomial Regression: Displacement vs MPG')

plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    results_df['Degree'],
    results_df['R-squared'],
    marker='o'
    
)

plt.xlabel("Polynomial Degree")
plt.ylabel("R-squared")
plt.title("Comparison of R-squared")
```
